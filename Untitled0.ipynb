Bir cümlede belirsizlik çözümünü analiz edip uygulamaya yönelik bir örnek verilmiştir.

```python
import pandas as pd

# 1. Java'daki 'createWithDefaults()' metodu
# TurkishMorphology nesnesinin oluşturulması
""" Bu modül, temel Türk morfolojik analizi ve üretimi sağlar. Analiz kelime ve
cümle düzeyinde yapılabilir. Kelime düzeyinde analiz ve üretim için TurkishMorphology
sınıfı kullanılır. Varsayılan olarak, sınıf şu şekilde oluşturulur: """

morphology = TurkishMorphology.create_with_defaults()

sentence = """
Bolu'da çam ormanları arasında saklı Gölcük Tabiat Parkı’nda sonbaharla birlikte renk cümbüşü yaşanıyor. Gölcük’te kartpostallık görüntüler oluştu.


Sonbaharla birlikte renk cümbüşünün yaşandığı tabiat parkını görenler hayran kalıyor.


Yeşil ve sarı ile mavinin birleştiği parkta kartpostallık görüntüler oluştu.


Gölün etrafında yürüyüşe çıkanlar, doğa harikasının keyfini çıkardı.
"""

# 2. Java'daki 'analyzeSentence(sentence)' metodu
# Python wrapper'ında bu genellikle 'analyze_sentence' veya 'analyze' olarak sarmalanır
try:
    analysis_result = morphology.analyze_sentence(sentence)
except AttributeError:
    # Eğer cümleniz hazır bölünmüşse tekil analiz de kullanılabilir
    analysis_result = [morphology.analyze(w) for w in sentence.split()]

print("--- Belirsizlik Öncesi (Adayların Çıkarılması) ---")
dataset_rows = []

for word_analysis in analysis_result:
    # Java'daki 'entry.getInput()' karşılığı
    try:
        word = word_analysis.getInput()
    except AttributeError:
        word = getattr(word_analysis, 'input', "UNK")

    print(f"\nWord = {word}")

    # Java'daki 'for (SingleAnalysis single : entry)' döngüsü
    try:
        candidates = list(word_analysis.all_matches)
    except AttributeError:
        candidates = list(word_analysis)

    for single in candidates:
        # Java'daki 'single.formatLong()' metodu
        # Python'da str() ile objenin string temsilini alıyoruz.
        analysis_str = str(single)
        print(f"  Aday Analiz: {analysis_str}")

        # Bu adayları makine öğrenmesi satırlarına dönüştürüyoruz
        dataset_rows.append({
            "Kelime": word,
            "Zemberek_Aday_String": analysis_str,
            # Buraya daha sonra CoNLL dosyasındaki 'doğru cevap' ile eşleşme (0 veya 1) gelecek
            "Hedef_Etiket": 0
        })

# 3. DataFrame yapısına aktarma
df = pd.DataFrame(dataset_rows)
```

```python
!pip install sklearn-crfsuite
import sklearn_crfsuite
from sklearn_crfsuite import metrics
from sklearn.model_selection import KFold
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix

# 1. Veri Okuma
def load_conll_data(file_path):
    dataset, current_sent = [], []
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line or line.startswith('#'): continue

            parts = line.split('\t')
            # 1. Kelimeyi al (parts[1])
            word = parts[1]

            # 2. Etiketi temizle (parts[3])
            # Sadece büyük harf olan, temel etiketi almak için:
            raw_label = parts[3]
            clean_label = raw_label.split()[0].upper() # "NOUN Case=..." kısmını "NOUN" yapar

            current_sent.append((word, clean_label))
    return dataset

# 2. Özellik Mühendisliği (Lisans bilgine uygun yapı)
def word2features(sent, i):
    word = sent[i][0]
    return {
        'bias': 1.0, 'word.lower()': word.lower(),
        'word[-3:]': word[-3:], 'is_first': i == 0,
        'is_last': i == len(sent) - 1, 'is_capitalized': word[0].isupper() if word else False
    }

# 3. Eğitim ve Raporlama Hattı
dataset = load_dataset('/content/drive/MyDrive/Colab_Notebooks_YL/dogal_dil_isleme/dataset/dataset.conll')
X = [[word2features(s, i) for i in range(len(s))] for s in dataset]
y = [[label for token, label in s] for s in dataset]

crf = sklearn_crfsuite.CRF(algorithm='lbfgs', c1=0.1, c2=0.1, all_possible_transitions=True)

# 5-Fold Cross-Validation
kf = KFold(n_splits=5, shuffle=True, random_state=42)
for train_idx, test_idx in kf.split(X):
    X_train, X_test = [X[i] for i in train_idx], [X[i] for i in test_idx]
    y_train, y_test = [y[i] for i in train_idx], [y[i] for i in test_idx]

    crf.fit(X_train, y_train)
    y_pred = crf.predict(X_test)

    # Başarı Oranları (Raporun için)
    print(metrics.flat_classification_report(y_test, y_pred, digits=3))

    # Karışıklık Matrisi (Confusion Matrix)
    # y_test ve y_pred listeler listesi olduğu için düzleştirmemiz gerekiyor
    y_test_flat = [label for sublist in y_test for label in sublist]
    y_pred_flat = [label for sublist in y_pred for label in sublist]

    cm = confusion_matrix(y_test_flat, y_pred_flat, labels=list(crf.classes_))
    sns.heatmap(cm, annot=True, fmt='d', xticklabels=crf.classes_, yticklabels=crf.classes_)
    plt.show()
```
